Case Study– Departmental Data Tracker
Instructions:

Create a dictionary department with keys: "dept_name", "faculty", "courses", "budget".

Add and modify items using dictionary methods (update, pop, get, etc.).

Create two sets: online_courses and offline_courses.

Perform union, intersection, difference, and symmetric_difference.
Create a frozenset named permanent_faculty.

```python
# Step 1: Department Dictionary
department={"dept_name":"Bio_Math","faculty":"Meera","courses":("Maths","Bio","Chemistry"),"budget":10000}
print(department)
```

```python
# Step 2: Modify values
department.update({"Duration":"3 Months"})
print(department)
department.pop("Duration")
print(department)
department.update({"Duration":"3 Months"})
print(department)
department.popitem()
print(department)
department.get("dept_name")
print(department.keys())
print(department.values())

department1=department.copy()
print(department1)
department1.clear()
print(department1)
```

```python
# Step 3: Set Operations
online_courses={"AI","ML","DA","DS","SQL","PYTHON"}
offline_courses={"ML","DS","PYTHON","C++","JAVA","EXCEL"}
print("online_courses=",online_courses)
print("offline_courses=",offline_courses)
print("union=",online_courses.union(offline_courses))
print("intersection=",online_courses.intersection(offline_courses))
print("difference=",online_courses.difference(offline_courses))
print("difference=",offline_courses.difference(online_courses))
print("symmetric_difference=",online_courses.symmetric_difference(offline_courses))
```

```python
# Step 4: Frozenset Example
Faculty={"Meera","Geetha","Anitha","Saranya"}
print(Faculty)

permanent_faculty=frozenset(Faculty)
print("permanent_faculty: ",permanent_faculty)
```
