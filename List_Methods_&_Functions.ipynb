```python
## Create three lists of course enrollments (Python, SQL, Power BI).

python= [50,48,37,39]
sql= [35,40,45,50,55]
powerbi= [25,29,33,37,41,45]

## Use append(), insert(), and extend() to update them.

python.append(42)
sql.insert(1,38)
powerbi.extend([49,53,57])
print(python)
print(sql)
print(powerbi)
```

```python
python.pop()
print(python)
sql.pop(3)
print(sql)
powerbi.remove(37)
print(powerbi)
```

```python
sql1=sql.copy()
print(sql)
print(sql1)
sql1.clear()
print(sql)
print(sql1)
```

```python
# Display the first two and last two enrollments using slicing.
sql[:2]
```

```python
powerbi[-2:]
```

```python
# Find total, max, and min enrollments using built-in functions.

max(sql)
```

```python
min(python)
```

```python
sum(powerbi)
```

```python
# Sort the lists in ascending and descending order.
sql.sort()
print(sql)
```
